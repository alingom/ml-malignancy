```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt
 
ds = load_breast_cancer()
X = pd.DataFrame(ds.data, columns=ds.feature_names)
y = pd.Series(ds.target, name='target')
print(X.shape, y.value_counts())
X.describe().T.head()
```

```python
# Simple histogram for a key feature
X['mean radius'].hist(bins=30)
plt.title('Distribution: mean radius')
plt.xlabel('mean radius'); plt.ylabel('count')
plt.show()
```
